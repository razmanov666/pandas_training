```python
import pandas as pd
```

```python
# reading the csv file
df_population_raw = pd.read_csv('population_total.csv')
```

```python
# first look at the dataset
df_population_raw
```

# Making a Pivot Table

.pivot(): Returns reshaped DataFrame organized by given index / column values ("pivot without aggregation")

```python
# dropping null values
df_population_raw.dropna(inplace=True)
```

```python
# making a pivot table
df_population_table = df_population_raw.pivot_table(index='year', columns="country", values="population")
df_population_table
```

```python
# selecting some countries
df_pivot = df_population_table[["Russia", "India", "China", "Nigeria", "Indonesia"]]
```

# Lineplot

```python
df_pivot.plot(kind='line', xlabel='Year', ylabel='Population', title='Population (1955-2020)', figsize=(8,4))
```

# Barplot

## Single Barplot

```python
# selecting only one year (2020)
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020.0])]
```

```python
# showing new dataframe
df_pivot_2020
```

```python
# transpose dataframe from rows to columns or viceversa
df_pivot_2020 = df_pivot_2020.T
```

```python
# making barplot
df_pivot_2020.plot(kind='bar', color='orange')
```

## Barplot grouped by “n” variables

```python
# selecting a few years
df_pivot_sample = df_pivot[df_pivot.index.isin([1980, 1990, 2000, 2010, 2020.0])]
```

```python
# making grouped barplot
df_pivot_sample.plot(kind='bar', figsize=(10,10))
```
