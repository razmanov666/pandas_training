```python
import pandas as pd
from IPython.display import display, HTML
import cufflinks as cf

cf.set_config_file(
    sharing='public',
    theme='black',
    offline=True
)
```

```python
# reading the csv file
df_population_raw = pd.read_csv('population_total.csv')
# dropping null values
df_population_raw.dropna(inplace=True)
# making a pivot table
df_pivot = df_population_raw.pivot(index='year', columns='country',
                                        values='population')
# selecting some countries
# df_pivot = df_pivot[['United States', 'India', 'China', 
#                           'Indonesia', 'Brazil']]
# showing pivot table
df_pivot

import plotly.express as px

# fig = px.line(
#     df_pivot,
#     x=df_pivot.index,
#     y=df_pivot.columns,
#     title='Population Growth by Country'
# )

# fig.update_layout(
#     xaxis_title='Year',
#     yaxis_title='Population',
#     legend_title='Country',
#     template='plotly_white'
# )

fig = px.scatter(
    df_pivot,
    title='Population Growth by Country'
)

fig.update_layout(
    xaxis_title='Country',
    yaxis_title='Population',
    legend_title='Country',
    template='seaborn'
)

fig.show()
```

```python
# selecting only one year (2020)
# transpose dataframe from rows to columns or viceversa
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
df_pivot_2020 = df_pivot_2020.T

fig = px.bar(
    df_pivot_2020,
    y=df_pivot_2020.index,
    x=df_pivot_2020.columns,
    title='Population Growth by Country'
)
```
