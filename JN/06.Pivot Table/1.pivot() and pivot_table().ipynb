# pivot()

```python
# read and show the dataset
df_gdp = pd.read_csv('gdp.csv', encoding='windows-1251')
df_gdp
```

```python
# reshape the dataframe with .pivot() 
df_gdp_pivot = df_gdp.pivot(index="country", columns="year", values="gdppc")
# df_gdp_pivot.loc['Russia']
df_gdp_pivot.head(50)
```

# pivot_table()

```python
# read and show the dataset
df_sales = pd.read_excel("supermarket_sales.xlsx")
df_sales
```

```python
# make a pivot table and add an aggregate function

# select_dtypes(include='number') автоматически выберет все инты и флоаты
numeric_cols = df_sales.select_dtypes(include='number').columns

# Строим pivot_table только по нужным столбцам
df_sales[[*numeric_cols, 'Gender']].pivot_table(index='Gender', aggfunc='sum')
```

```python
# make a pivot table, add an aggregate function and select some columns 
df_sales.pivot_table(index='Gender', 
                     values=['Quantity', 'Total'],
                     aggfunc='sum')
```

```python
# make a pivot tables that says how much male and female spend in each category
df_sales.pivot_table(index='Gender',
                     columns="Product line",
                     values='Total',
                     aggfunc='sum')
```
